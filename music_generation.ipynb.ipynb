```python
from music21 import converter, note, chord
import numpy as np

notes = []

# Start with 20 MIDI files
for file in midi_files[:20]:

    print("Processing:", file)

    try:
        midi = converter.parse(file)

        for element in midi.flatten().notes:

            # Individual note
            if isinstance(element, note.Note):
                notes.append(str(element.pitch))

            # Chord
            elif isinstance(element, chord.Chord):
                notes.append(
                    '.'.join(str(n) for n in element.normalOrder)
                )

    except Exception as e:
        print("Error processing:", file)
        print(e)

print("\nTotal musical events:", len(notes))
print("\nFirst 30 events:")
print(notes[:30])
```

```python
print("Number of musical events:", len(notes))
print("Number of unique events:", len(set(notes)))

print("\nSome unique events:")
print(list(set(notes))[:30])
```

```python
# Create a sorted list of all unique musical events
pitchnames = sorted(set(notes))

# Create a dictionary:
# musical event -> number
note_to_int = dict(
    (note, number)
    for number, note in enumerate(pitchnames)
)

print("Number of unique events:", len(pitchnames))

print("\nFirst 20 mappings:")
for item in pitchnames[:20]:
    print(item, "->", note_to_int[item])
```

```python
sequence_length = 40

input_sequences = []
output_notes = []

for i in range(len(notes) - sequence_length):

    sequence = notes[i:i + sequence_length]
    next_note = notes[i + sequence_length]

    input_sequences.append(
        [note_to_int[n] for n in sequence]
    )

    output_notes.append(
        note_to_int[next_note]
    )

print("Number of input sequences:", len(input_sequences))
print("Length of each sequence:", len(input_sequences[0]))
print("Number of output notes:", len(output_notes))
```

```python
print("First input sequence:")
print(input_sequences[0])

print("\nFirst expected output:")
print(output_notes[0])
```

```python
import numpy as np

X = np.array(input_sequences)
y = np.array(output_notes)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X = np.reshape(
    X,
    (X.shape[0], X.shape[1], 1)
)

print("New X shape:", X.shape)
```

```python
X = X / float(len(pitchnames))

print("Minimum value in X:", X.min())
print("Maximum value in X:", X.max())
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dropout, Dense

model = Sequential()

model.add(
    LSTM(
        256,
        input_shape=(X.shape[1], X.shape[2]),
        return_sequences=True
    )
)

model.add(Dropout(0.3))

model.add(
    LSTM(256)
)

model.add(Dropout(0.3))

model.add(
    Dense(len(pitchnames), activation="softmax")
)

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam"
)

model.summary()
```

```python
import tensorflow as tf

print("Available GPUs:")
print(tf.config.list_physical_devices("GPU"))
```

```python
history = model.fit(
    X,
    y,
    epochs=1,
    batch_size=64,
    verbose=1
)
```

```python
from music21 import converter, note, chord
import numpy as np
import pickle
import os

notes = []

for file in midi_files[:20]:

    print("Processing:", os.path.basename(file))

    try:
        midi = converter.parse(file)

        for element in midi.flatten().notes:

            if isinstance(element, note.Note):
                notes.append(str(element.pitch))

            elif isinstance(element, chord.Chord):
                notes.append(
                    '.'.join(str(n) for n in element.normalOrder)
                )

    except Exception as e:
        print("Error processing:", file)
        print(e)

print("\nTotal musical events:", len(notes))
print("Unique musical events:", len(set(notes)))
```

```python
pitchnames = sorted(set(notes))

note_to_int = {
    item: number
    for number, item in enumerate(pitchnames)
}

print("Number of unique events:", len(pitchnames))
```

```python
sequence_length = 40

input_sequences = []
output_notes = []

for i in range(len(notes) - sequence_length):

    sequence = notes[i:i + sequence_length]
    next_note = notes[i + sequence_length]

    input_sequences.append(
        [note_to_int[n] for n in sequence]
    )

    output_notes.append(
        note_to_int[next_note]
    )

X = np.array(input_sequences)
y = np.array(output_notes)

X = np.reshape(
    X,
    (X.shape[0], X.shape[1], 1)
)

X = X / float(len(pitchnames))

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Data preparation complete!")
```

```python
import numpy as np

start = np.random.randint(
    0,
    len(X)
)

pattern = X[start].flatten().tolist()

# Convert normalized values back to integer event IDs
pattern = [
    int(value * len(pitchnames))
    for value in pattern
]

print("Starting sequence length:", len(pattern))
print("Starting sequence:")
print(pattern[:40])
```

```python
import numpy as np

prediction_output = []

# Generate 200 new events
for _ in range(200):
    # Prepare input for the model
    input_pattern = np.array(pattern).reshape(1, len(pattern), 1)
    input_pattern = input_pattern / float(len(pitchnames))

    # Predict the next event
    prediction = model.predict(input_pattern, verbose=0)

    # Choose the most likely event
    index = np.argmax(prediction)

    # Add predicted event
    result = pitchnames[index]
    prediction_output.append(result)

    # Move sequence forward
    pattern.append(index)
    pattern = pattern[1:]

print("Generated events:", len(prediction_output))
print("First 20 generated events:")
print(prediction_output[:20])
```

```python
import numpy as np

# Start again from a random sequence
start = np.random.randint(0, len(X))

pattern = X[start].flatten().tolist()

# Convert normalized values back to event IDs
pattern = [
    min(int(value * len(pitchnames)), len(pitchnames) - 1)
    for value in pattern
]

prediction_output = []

# Generate 200 events
for _ in range(200):

    # Prepare input
    input_pattern = np.array(pattern).reshape(1, len(pattern), 1)
    input_pattern = input_pattern / float(len(pitchnames))

    # Predict probabilities
    prediction = model.predict(input_pattern, verbose=0)[0]

    # Sample from the probability distribution
    index = np.random.choice(
        len(prediction),
        p=prediction / prediction.sum()
    )

    # Save generated event
    prediction_output.append(pitchnames[index])

    # Update pattern
    pattern.append(index)
    pattern = pattern[1:]

print("Generated events:", len(prediction_output))
print("First 20 generated events:")
print(prediction_output[:20])
```

```python
import numpy as np

# Start again from a random sequence
start = np.random.randint(0, len(X))

pattern = X[start].flatten().tolist()

# Convert normalized values back to event IDs
pattern = [
    min(int(value * len(pitchnames)), len(pitchnames) - 1)
    for value in pattern
]

prediction_output = []

# Generate 200 events
for _ in range(200):

    # Prepare input
    input_pattern = np.array(pattern).reshape(1, len(pattern), 1)
    input_pattern = input_pattern / float(len(pitchnames))

    # Predict probabilities
    prediction = model.predict(input_pattern, verbose=0)[0]

    # Sample from the probability distribution
    index = np.random.choice(
        len(prediction),
        p=prediction / prediction.sum()
    )

    # Save generated event
    prediction_output.append(pitchnames[index])

    # Update pattern
    pattern.append(index)
    pattern = pattern[1:]

print("Generated events:", len(prediction_output))
print("First 20 generated events:")
print(prediction_output[:20])
```

```python
from music21 import stream, note, chord

generated_stream = stream.Stream()

for current_event in prediction_output:

    # If the event contains a dot, treat it as a chord
    if "." in current_event:
        pitch_classes = current_event.split(".")

        chord_notes = []
        for pc in pitch_classes:
            try:
                midi_number = 60 + int(pc)
                chord_notes.append(note.Note(midi=midi_number))
            except:
                pass

        if chord_notes:
            new_chord = chord.Chord(chord_notes)
            new_chord.quarterLength = 0.5
            generated_stream.append(new_chord)

    # Otherwise, treat it as a single note
    else:
        try:
            new_note = note.Note(current_event)
            new_note.quarterLength = 0.5
            generated_stream.append(new_note)
        except:
            pass

# Save MIDI file
output_midi_path = "/content/drive/MyDrive/CodeAlpha_MusicGeneration/generated_music.mid"

generated_stream.write("midi", fp=output_midi_path)

print("MIDI file created successfully!")
print("Saved to:")
print(output_midi_path)
print("Number of musical elements:", len(generated_stream))
```
